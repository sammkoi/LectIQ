```python
import pandas as pd
import re
```

```python
all_sheets = pd.read_excel("galectins_id.xlsx", sheet_name=None)
data = {}
for glectin in all_sheets:
  temp = all_sheets[glectin]
  data[glectin] = temp[temp["Glycan"].notna()].reset_index(drop=True)
  data[glectin] = data[glectin].drop(columns=['Symbol'])
```

```python
with pd.ExcelWriter("galectins_id_cleaned.xlsx") as writer:
  for glectin, df in data.items():
    df.to_excel(writer, sheet_name=glectin, index=False)
```

```python
with pd.ExcelWriter("output.xlsx") as writer:
  for glectin in data:
    df = data[glectin]
    df.to_excel(writer, sheet_name=re.sub(r'[\\/*?:[\]]', '_', glectin), index=False)
```
