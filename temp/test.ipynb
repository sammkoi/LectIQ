```python
import re

def convert_to_iupac(glycan_str):
    """
    Converts a glycan string like 
    'β-D-Gal-(1→4)-β-D-GlcNAc-(1→3)-β-D-Gal'
    into condensed IUPAC format like
    'Galβ1-4GlcNAcβ1-3Gal'
    """
    # Remove whitespace
    glycan_str = glycan_str.replace(" ", "")
    
    # Replace unicode arrow with standard
    glycan_str = glycan_str.replace("→", "-")
    
    # Regex to match each sugar unit with optional branch
    # e.g. β-D-GlcNAc-(1-3) or [α-D-Man-(1-6)]
    sugar_pattern = re.compile(r'(\[?[αβ]-D-[A-Za-z0-9]+-?\(?[0-9-]*\)?\]?)')
    matches = sugar_pattern.findall(glycan_str)
    
    condensed = ""
    for unit in matches:
        # Remove α/β and D- prefix
        unit_clean = re.sub(r'[\[αβ]-D-', '', unit)
        unit_clean = unit_clean.replace(']', '')
        
        # Extract linkage
        linkage_match = re.search(r'\(?([0-9-]+)\)?', unit)
        linkage = linkage_match.group(1) if linkage_match else ""
        if linkage:
            linkage = linkage.replace("-", "")
            unit_clean += linkage
        
        # Add '-' if not at the end
        if condensed:
            condensed += "-"
        condensed += unit_clean
    
    return condensed

# Example usage
# glycan = "β-D-Gal-(1→4)-β-D-GlcNAc-(1→3)-β-D-Gal-(1→4)-β-D-GlcNAc-(1→3)-β-D-Gal-(1→4)-β-D-Glc"
gc=temp["Glycan"][10]
iupac = convert_to_iupac(gc)
print(iupac)
```
